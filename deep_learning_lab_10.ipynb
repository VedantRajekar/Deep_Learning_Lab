### 3. Data Preprocessing and Feature Engineering

We need to prepare the data for the deep neural network. This involves:
1.  **Dropping irrelevant columns**: Removing identifiers or columns that don't add predictive power.
2.  **Feature Engineering**: Extracting useful information from existing columns (e.g., time features).
3.  **Handling Categorical Features**: Converting categorical data into a numerical format suitable for neural networks.
4.  **Scaling Numerical Features**: Normalizing numerical features to a similar range to help the neural network converge faster.

```python
import numpy as np

# Drop irrelevant columns
columns_to_drop = [
    'Unnamed: 0', # Likely an artifact/index column
    'cc_num',     # Credit card number (identifier, sensitive, high cardinality)
    'first',      # First name (identifier)
    'last',       # Last name (identifier)
    'street',     # Street address (high cardinality, location already covered by lat/long)
    'trans_num',  # Transaction number (identifier)
    'dob',        # Date of birth (can lead to privacy concerns and high cardinality if not engineered carefully, simpler to drop for initial model)
    'city',       # City (high cardinality, consider dropping for initial model)
    'state'       # State (high cardinality, consider dropping for initial model)
]

# Check if columns exist before dropping
existing_columns_to_drop = [col for col in columns_to_drop if col in df.columns]

if existing_columns_to_drop:
    df_processed = df.drop(columns=existing_columns_to_drop)
    print(f"Dropped columns: {existing_columns_to_drop}")
else:
    df_processed = df.copy()
    print("No specified columns to drop were found in the DataFrame.")

# Convert 'trans_date_trans_time' to datetime objects and extract features
df_processed['trans_date_trans_time'] = pd.to_datetime(df_processed['trans_date_trans_time'])

df_processed['hour'] = df_processed['trans_date_trans_time'].dt.hour
df_processed['day'] = df_processed['trans_date_trans_time'].dt.day
df_processed['month'] = df_processed['trans_date_trans_time'].dt.month
df_processed['year'] = df_processed['trans_date_trans_time'].dt.year
df_processed['day_of_week'] = df_processed['trans_date_trans_time'].dt.dayofweek
df_processed['is_weekend'] = (df_processed['trans_date_trans_time'].dt.dayofweek >= 5).astype(int) # Saturday=5, Sunday=6

# Drop the original 'trans_date_trans_time' and 'unix_time' after extraction
df_processed = df_processed.drop(columns=['trans_date_trans_time', 'unix_time'])

print("\nDataFrame after initial drops and feature engineering:")
display(df_processed.head())
df_processed.info()
```

Next, we'll handle categorical features using one-hot encoding and scale the numerical features. This is crucial for deep learning models.

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Identify numerical and categorical features
numerical_features = df_processed.select_dtypes(include=np.number).columns.tolist()
categorical_features = df_processed.select_dtypes(include='object').columns.tolist()

# Remove the target variable 'is_fraud' from numerical features list
if 'is_fraud' in numerical_features:
    numerical_features.remove('is_fraud')

print(f"Numerical features: {numerical_features}")
print(f"Categorical features: {categorical_features}")

# Create preprocessing pipelines for numerical and categorical features
numerical_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown='ignore') # handle_unknown='ignore' for unseen categories in test set

# Create a column transformer to apply different transformations to different columns
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='passthrough' # Keep any other columns (e.g., 'is_fraud' which was not in numerical or categorical list)
)

# Apply the preprocessing steps
# Temporarily separate target variable before applying preprocessor
X = df_processed.drop(columns=['is_fraud'])
y = df_processed['is_fraud']

# Fit and transform the features
X_processed = preprocessor.fit_transform(X)

# Convert the processed features back to a DataFrame for easier handling, if needed
# This step can be memory intensive if many new columns are created by OneHotEncoder
# For DNN, often directly use the numpy array X_processed
print("\nShape of processed features (X_processed):", X_processed.shape)
print("Shape of target variable (y):", y.shape)

# To see column names after one-hot encoding, you can reconstruct DataFrame (optional)
# feature_names = numerical_features + list(preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features))
# X_processed_df = pd.DataFrame(X_processed, columns=feature_names)
# display(X_processed_df.head())
```

### 4. Splitting Data and Applying SMOTE

Before training our deep neural network, we need to:
1.  **Split the data**: Separate the dataset into training and testing sets.
2.  **Apply SMOTE**: Use SMOTE (Synthetic Minority Over-sampling Technique) on the *training data only* to address the class imbalance. This will generate synthetic samples for the minority class, helping the model learn better.

```python
# Install imbalanced-learn for SMOTE
!pip install imbalanced-learn

from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
# Stratify by 'y' to maintain the same class distribution in both sets
# Use a test size of 20% (0.2), which is common
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42, stratify=y)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")

print("\nClass distribution in y_train before SMOTE:")
display(y_train.value_counts())
display(y_train.value_counts(normalize=True) * 100)
```

```python
from imblearn.over_sampling import SMOTE

print("Applying SMOTE to the training data...")

smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print("SMOTE application complete.")
print(f"Shape of X_train_resampled: {X_train_resampled.shape}")
print(f"Shape of y_train_resampled: {y_train_resampled.shape}")

print("\nClass distribution in y_train after SMOTE:")
display(y_train_resampled.value_counts())
display(y_train_resampled.value_counts(normalize=True) * 100)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define the Deep Neural Network architecture
model = keras.Sequential([
    layers.Input(shape=(X_train_resampled.shape[1],)), # Input layer matching the number of features
    layers.Dense(256, activation='relu'), # First hidden layer with 256 neurons and ReLU activation
    layers.BatchNormalization(),          # Batch Normalization for stability and faster training
    layers.Dropout(0.3),                  # Dropout for regularization to prevent overfitting
    layers.Dense(128, activation='relu'), # Second hidden layer with 128 neurons and ReLU activation
    layers.BatchNormalization(),
    layers.Dropout(0.3),
    layers.Dense(64, activation='relu'),  # Third hidden layer with 64 neurons and ReLU activation
    layers.BatchNormalization(),
    layers.Dropout(0.2),
    layers.Dense(1, activation='sigmoid') # Output layer with 1 neuron and sigmoid for binary classification
])

# Compile the model
# Using Adam optimizer, Binary Crossentropy loss for binary classification, and F1-score as a metric
# F1-score is particularly useful for imbalanced datasets
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

# Display model summary
print("Model Summary:")
model.summary()

# Define callbacks for early stopping and learning rate reduction
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,             # Stop if no improvement after 5 epochs
    restore_best_weights=True
)
lr_scheduler = keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,             # Reduce learning rate by a factor of 0.2
    patience=3,             # If no improvement for 3 epochs
    min_lr=0.0001
)

print("\nTraining the model...")
# Train the model with the resampled training data and validation on the original test set
history = model.fit(
    X_train_resampled, y_train_resampled,
    validation_data=(X_test, y_test),
    epochs=20, # Start with a reasonable number of epochs, early stopping will prevent overtrianing
    batch_size=2048, # Larger batch size for efficiency with a large dataset
    callbacks=[early_stopping, lr_scheduler],
    verbose=1
)

print("\nModel training complete.")
```

### 5. Building and Training the Deep Neural Network Model

Now, we'll construct a deep neural network using TensorFlow/Keras. The model will be trained on the SMOTE-resampled training data to ensure it learns effectively from both fraud and non-fraud instances. We'll use appropriate layers and activation functions for a binary classification task.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define the Deep Neural Network architecture
model = keras.Sequential([
    layers.Input(shape=(X_train_resampled.shape[1],)), # Input layer matching the number of features
    layers.Dense(256, activation='relu'), # First hidden layer with 256 neurons and ReLU activation
    layers.BatchNormalization(),          # Batch Normalization for stability and faster training
    layers.Dropout(0.3),                  # Dropout for regularization to prevent overfitting
    layers.Dense(128, activation='relu'), # Second hidden layer with 128 neurons and ReLU activation
    layers.BatchNormalization(),
    layers.Dropout(0.3),
    layers.Dense(64, activation='relu'),  # Third hidden layer with 64 neurons and ReLU activation
    layers.BatchNormalization(),
    layers.Dropout(0.2),
    layers.Dense(1, activation='sigmoid') # Output layer with 1 neuron and sigmoid for binary classification
])

# Compile the model
# Using Adam optimizer, Binary Crossentropy loss for binary classification, and F1-score as a metric
# F1-score is particularly useful for imbalanced datasets
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

# Display model summary
print("Model Summary:")
model.summary()

# Define callbacks for early stopping and learning rate reduction
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,             # Stop if no improvement after 5 epochs
    restore_best_weights=True
)
lr_scheduler = keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,             # Reduce learning rate by a factor of 0.2
    patience=3,             # If no improvement for 3 epochs
    min_lr=0.0001
)

print("\nTraining the model...")
# Train the model with the resampled training data and validation on the original test set
history = model.fit(
    X_train_resampled, y_train_resampled,
    validation_data=(X_test, y_test),
    epochs=5, # Start with a reasonable number of epochs, early stopping will prevent overtrianing
    batch_size=2048, # Larger batch size for efficiency with a large dataset
    callbacks=[early_stopping, lr_scheduler],
    verbose=1
)

print("\nModel training complete.")
```

### 6. Model Evaluation

After training, it's important to evaluate the model's performance on the unseen test set (`X_test`, `y_test`). We will look at key metrics relevant to imbalanced classification, such as accuracy, precision, and recall.

```python
from sklearn.metrics import classification_report, confusion_matrix

print("\nEvaluating the model on the test set...")
loss, accuracy, precision, recall = model.evaluate(X_test, y_test, verbose=0)

print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
print(f"Test Precision: {precision:.4f}")
print(f"Test Recall: {recall:.4f}")

# Predict probabilities on the test set
y_pred_prob = model.predict(X_test)
# Convert probabilities to binary predictions (0 or 1) using a threshold (e.g., 0.5)
y_pred = (y_pred_prob > 0.5).astype(int)

print("\nConfusion Matrix:")
display(pd.DataFrame(confusion_matrix(y_test, y_pred), index=['Actual 0', 'Actual 1'], columns=['Predicted 0', 'Predicted 1']))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

### 7. Discussion and Next Steps

Based on the evaluation metrics, we can now assess the performance of our deep neural network in detecting fraudulent transactions. The precision, recall, and F1-score are particularly important for understanding how well the model identifies fraud while minimizing false positives.

**Key Observations:**
*   **Precision:** Indicates the proportion of positive identifications that were actually correct. A high precision means fewer legitimate transactions are flagged as fraudulent.
*   **Recall:** Indicates the proportion of actual positives that were correctly identified. A high recall means the model catches most fraudulent transactions.
*   **F1-Score:** The harmonic mean of precision and recall, providing a balanced measure.
